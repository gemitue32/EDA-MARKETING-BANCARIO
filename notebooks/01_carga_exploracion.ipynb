# 01 · Carga y exploración inicial

Carga de los dos datasets (`bank-additional.csv` y `customer-details.xlsx`) y diagnóstico de su calidad: columnas, tipos, nulos, duplicados y formato de los valores. **En este notebook no se modifica ningún dato.**

```python
# Rutas y carga de los datos

import pandas as pd

ruta_bank = "../data/raw/bank-additional.csv"
ruta_customer = "../data/raw/customer-details.xlsx"
```

```python
df_bank = pd.read_csv(ruta_bank)
hojas_customer = pd.read_excel(ruta_customer, sheet_name=None)

print(type(df_bank))
print(type(hojas_customer))
```

```python
# Tamaño de los datos

for nombre, hoja in hojas_customer.items():
    print(nombre, hoja.shape)
print(df_bank.shape)

total_customer = sum(hoja.shape[0] for hoja in hojas_customer.values())
print(total_customer)

```

```python
# Columnas: comparar con el enunciado

print(hojas_customer["2012"].columns)
print(df_bank.columns)
```

```python
# Revisar las coordenadas

print(df_bank[["latitude", "longitude"]].describe())
```

```python
# Formato de la fecha

print(df_bank["date"].head(10))
```

```python
# Tipos de dato

df_bank.info()
```

```python
# Nulos

nulos = df_bank.isnull().sum()
print(nulos[nulos > 0])

porc_nulos = (nulos[nulos > 0] / len(df_bank) * 100).round(2)
print(porc_nulos.sort_values(ascending=False))
```

```python
# Números leídos como texto

cols_economicas = ["emp.var.rate", "cons.price.idx", "cons.conf.idx", "euribor3m", "nr.employed"]
print(df_bank[cols_economicas].head())
print(df_bank["nr.employed"].unique())
```

```python
# Cómo están escritas las categorías

cols_texto = ["job", "marital", "education", "contact", "poutcome", "y"]
for col in cols_texto:
    print(df_bank[col].value_counts(dropna=False))
    print("-" * 40)
```

```python
# Duplicados y revisión del Excel

print("Filas duplicadas:", df_bank.duplicated().sum())
print("IDs duplicados:", df_bank["id_"].duplicated().sum())

for nombre, hoja in hojas_customer.items():
    print(nombre)
    print("Nulos:", hoja.isnull().sum().sum())
    print(hoja.dtypes)
    print("-" * 40)
```

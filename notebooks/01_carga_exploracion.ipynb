## Diario de hallazgos

| Dataset | Columna | Problema | Decisión |
|---|---|---|---|
| customer-details | Unnamed: 0 | Índice guardado como columna | Eliminar |
| customer-details | (filas) | 170 clientes más que llamadas | Comprobar al unir |
| customer-details | 3 hojas | Mismas columnas repartidas en 3 tablas | Unirlas con pd.concat sin perder el año |
| bank | Unnamed: 0 | Índice guardado como columna | Eliminar |
| bank | latitude, longitude | Coordenadas de EE. UU. (lat 24–49, lon -125 a -67), no de Portugal | Eliminar |
| bank | contact_month, contact_year | No existen | Crearlas a partir de date |
| bank | date | Texto, no fecha; mes escrito en español (2-agosto-2019) | Traducir meses a número y convertir a fecha |
| bank | cons.price.idx, cons.conf.idx, euribor3m, nr.employed | Leídas como texto: decimal con coma (93,994) | Cambiar coma por punto y convertir a número |
| bank | age, job, marital, education, default, housing, loan, cons.price.idx, euribor3m, date | Tienen nulos (de 0,2 % a 21,5 %) | Decidir según el % y el tipo de cada una |
| bank | default, housing, loan | Son 0/1 pero salen como float por los nulos | Pasar a entero tras tratar los nulos |
| bank | marital, poutcome | En mayúsculas; el resto en minúsculas | Pasar todo a minúsculas |
| bank | job, education | Puntos en los valores (admin., university.degree) | Limpiar puntos |
| bank | y | Texto yes/no | Convertir a 1/0 para calcular tasas |
